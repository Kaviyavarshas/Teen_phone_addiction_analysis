# DATA VISULASIZATION

# Univariate continuous data-histogram

```python
plt.hist(x='Daily_Usage_Hours',data=phone_addiction_df)
```

```python
plt.figure(figsize=(10,5))                                    
plt.hist(x='Sleep_Hours',data=phone_addiction_df)
plt.xlabel("Sleep_Hours")
plt.ylabel("Freq")
plt.title("Sleep_Hour Distribution",size=15)
plt.show()
```

# Univariate discrete data-CountPlot

```python
phone_addiction_df
```

```python
plt.bar(x="Age",data=phone_addiction_df)  
```

```python
sns.countplot(x="Age",data=phone_addiction_df)   
plt.show  
```

```python
purpose_counts=phone_addiction_df["Phone_Usage_Purpose"].value_counts()
plt.figure(figsize=(6,6))
plt.pie(purpose_counts,labels=purpose_counts.index,autopct='%1.1f%%')
plt.title("Phone Usage Distribution")
plt.show()
```

# Univariate continuous - boxplot

```python
plt.boxplot(x='Academic_Performance',data=phone_addiction_df) 
plt.show() 
```
